# 4) 🖼️ Tarefa 3 — Suavização com Filtro Passa-Baixas Ideal

Você deve:
- Transformar a imagem selecionada para o espectro das frequências.
- Aplicar na imagem um filtro passa-baixa, para tal utilizar a função ideal_lp_mask a seguir para criar a mascara (avaliar para diferentes valores de D0 - no mínimo três).
- Transformar o espectro de frequências filtrado de volta para o domínio espacial com a IFFT.
- Observar a imagem reconstruida no domínio espacial e discutir com o professor.

```python
def ideal_lp_mask(shape, D0):
    "Máscara passa-baixa IDEAL 2D (radial)."
    M, N = shape
    u = np.arange(M) - M/2
    v = np.arange(N) - N/2
    U, V = np.meshgrid(u, v, indexing='ij')
    D = np.sqrt(U**2 + V**2)
    # Frequências dentro do raio D0 passam (valor 1), fora são cortadas (valor 0)
    H = np.zeros_like(D)
    H[D <= D0] = 1.0
    return H
```
