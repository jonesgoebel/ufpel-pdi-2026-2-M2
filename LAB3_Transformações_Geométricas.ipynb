```python
# %% setup - Importações e funções utilitárias
import numpy as np
import cv2
import matplotlib.pyplot as plt

from PIL import Image
from pathlib import Path
from google.colab import files
```

```python
def converte_cinza(figura):
    """
    Recebe o caminho de uma imagem BMP, converte para escala de cinza, salva a imagem, retorna o caminho.
    """
    path = Path(figura)
    img = Image.open(figura).convert('L')

    # Salva a imagem convertida
    caminho_saida = path.with_name(f"{path.stem}_cinza{path.suffix}")
    img.save(caminho_saida)

    # Retorna o objeto Image
    return img
```

```python
def exibir_lado_a_lado(lista_figuras, lista_titulos):
    quantidade = len(lista_figuras)

    plt.figure(figsize=(5 * quantidade, 5))

    for i, img in enumerate(lista_figuras):

        plt.subplot(1, quantidade, i + 1)
        plt.imshow(img, cmap="gray")
        plt.title(lista_titulos[i])
        plt.axis("off")

    plt.show()
```

## 2) Transformação de Escala
A transformação de escala aumenta ou diminui o tamanho da imagem.
Vamos criar uma função **Python** que:
- Recebe o caminho de uma imagem em escala de cinza;
- Aplica transformação de escala usando matriz afim;
- Salva a imagem em BMP e retorna o caminho do arquivo gerado.

A matriz de escala usada será:

$$
\begin{bmatrix}
sx & 0 & 0 \\
0 & sy & 0
\end{bmatrix}
$$

Lembrando que:
* sx: fator de escala horizontal
* sy: fator de escala vertical

```python
def escala_img(img, sx, sy):
    arr = np.asarray(img, dtype=np.uint8)

    out = cv2.resize(
        arr,
        None,
        fx=sx,
        fy=sy,
        interpolation=cv2.INTER_LINEAR
    )

    return out
```

## 4) Transformação de Rotação
A rotação gira a imagem em torno de um ponto. Neste exemplo, vamos girar a imagem em torno do centro.

Vamos criar uma função **Python** que:
- Recebe o caminho de uma imagem em escala de cinza;
- Aplica rotação usando matriz afim;
- Salva a imagem em BMP e retorna o caminho do arquivo gerado.

* A rotação usa seno e cosseno:

$$
\begin{bmatrix}
\cos(\theta) & -\sin(\theta) & t_x \\
\sin(\theta) & \cos(\theta) & t_y
\end{bmatrix}
$$

```python
def rotacao_img(figura, angulo_graus):
    arr = np.asarray(figura, dtype=np.uint8)

    altura, largura = arr.shape

    centro = (largura / 2, altura / 2)

    matriz_rotacao = cv2.getRotationMatrix2D(
        centro,
        angulo_graus,
        1.0
    )

    out = cv2.warpAffine(
        arr,
        matriz_rotacao,
        (largura, altura)
    )

    return out
```

## 6) Translação
A translação desloca a imagem no eixo horizontal e no eixo vertical.

A matriz de translação é:

$$
\begin{bmatrix}
1 & 0 & t_x \\
0 & 1 & t_y
\end{bmatrix}
$$


```python
def translacao_opencv(figura, tx, ty):
    arr = np.asarray(figura, dtype=np.uint8)

    altura, largura = arr.shape

    # Matriz de translação
    matriz_translacao = np.float32([
        [1, 0, tx],
        [0, 1, ty]
    ])

    # Aplicando a transformação
    out = cv2.warpAffine(
        arr,
        matriz_translacao,
        (largura, altura)
    )

    return out
```

```python
img_cinza = converte_cinza("lena_orig.bmp")

trans_1 = translacao_opencv(img_cinza, 50, 0)
trans_2 = translacao_opencv(img_cinza, 0, 50)
trans_3 = translacao_opencv(img_cinza, 80, 80)
trans_4 = translacao_opencv(img_cinza, -50, -50)

exibir_lado_a_lado(
    [img_cinza, trans_1, trans_2, trans_3, trans_4],
    ["Original", "tx = 50", "ty = 50", "tx = 80, ty = 80", "tx = -50, ty = -50"]
)
```
