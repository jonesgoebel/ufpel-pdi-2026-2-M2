```python
# %% setup - Importações e funções utilitárias
import numpy as np
import cv2
import matplotlib.pyplot as plt
from skimage import io, color, data, img_as_ubyte
import imageio.v2 as imageio

from PIL import Image
from pathlib import Path
from google.colab import files
```

```python
def exibir_lado_a_lado(lista_figuras, lista_titulos):
    quantidade = len(lista_figuras)

    plt.figure(figsize=(5 * quantidade, 5))

    for i, img in enumerate(lista_figuras):

        plt.subplot(1, quantidade, i + 1)
        plt.imshow(img, cmap="gray")
        plt.title(lista_titulos[i])
        plt.axis("off")

    plt.show()

def mascara_retangular(h, w, x_inicio, y_inicio, x_fim, y_fim):
  mascara = np.zeros((h, w), dtype=np.uint8)
  cv2.rectangle(mascara, (x_inicio, y_inicio), (x_fim, y_fim), 255, -1)
  return mascara
```

## 2) Operação de Aritméticas (Subração)
As operações de subtração entre duas imagens podem ser utilizadas para fazer a detecção de mudanças.


```python
img1 = cv2.imread('/content/beirut_img1.png', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('/content/beirut_img2.png', cv2.IMREAD_GRAYSCALE)

print(f"Dimensão da Imagem {img1.shape}")

img_diff_sub = (np.abs(img1.astype(np.int16) - img2.astype(np.int16)))
img_diff = img_diff_sub.astype(np.uint8)

exibir_lado_a_lado([img1,img2,img_diff],["Imagem 1","Imagem 2","Diferença das Imagens"])
```
